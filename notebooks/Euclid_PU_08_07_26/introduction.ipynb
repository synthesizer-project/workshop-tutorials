# Introduction

This notebook provides a high-level overview of the `Synthesizer` package, which is a Python library for generating synthetic astronomical data. Full documentation is available the [docs](https://synthesizer-project.github.io/synthesizer/). 

`Synthesizer` is designed to be flexible and extensible, allowing users to create synthetic datasets for a wide range of applications in astronomy. The package includes modules for generating synthetic photometry, spectra, and images, as well as tools for adding realistic noise and other observational effects.

Synthesizer is designed to be agnostic to the source of data, and can load and process data from a variety of hydrodynamical simulations and semi-analytic models, and also includes a wide variety of parametric models for generating synthetic data from scratch.
